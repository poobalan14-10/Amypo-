```python
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns

data = {

"Hours_Studied": [2, 5, 3, 7, 1, 6, 4, 8, 5, 3],

"Attendance": [60, 90, 75, 95, 50, 88, 80, 98, 85, 70],

"Previous_Score": [45, 70, 55, 82, 40, 78, 65, 90, 72, 58],

"Final_Score": [42, 76, 58, 88, 35, 82, 68, 94, 79, 60]

}

df = pd.DataFrame(data)

print(df.head())

print("Shape:", df.shape)

print(df.isnull().sum())

sns.scatterplot(data=df, x="Hours_Studied", y="Final_Score")

plt.title("Study Hours vs Final Score")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Exact dataset specified
df = pd.DataFrame({
    "Student": ["Arun", "Priya", "Karthik", "Divya", "Vikram", "Ananya", "Suresh", "Meera", "Rahul", "Kavya"],
    "Hours_Studied": pd.Series([2, 5, 3, 7, 1, 6, 4, 8, 5, 3]),
    "Attendance": pd.Series([60, 90, 75, 95, 50, 88, 80, 98, 85, 70]),
    "Sleep_Hours": pd.Series([6.0, 7.5, 6.5, 8.0, 5.0, 7.0, None, 8.0, 8.5, 6.5]),
    "Final_Score": pd.Series([42, 76, 58, 88, 35, 82, 68, 150, 79, 60])
})

# Original Inspection
print(df.head())
print("Shape:", df.shape)
print(df.isnull().sum())

# Original Scatter Plot + New Box Plot
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 1. Original Scatter Plot
sns.scatterplot(data=df, x="Hours_Studied", y="Final_Score", ax=axes[0])
axes[0].set_title("Study Hours vs Final Score")

# 2. Box Plot (clearly highlights Meera's 150 outlier in Final_Score)
sns.boxplot(data=df[["Attendance", "Final_Score"]], ax=axes[1])
axes[1].set_title("Box Plot of Attendance & Final Score")

plt.tight_layout()
plt.show()
```
